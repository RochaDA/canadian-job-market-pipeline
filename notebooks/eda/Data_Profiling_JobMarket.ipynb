## Checking NULL ratios

```python
%sql
SELECT * 
FROM(
    SELECT
        ROUND(SUM(CASE WHEN job_posting_id IS NULL THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 1) AS job_posting_id,
        ROUND(SUM(CASE WHEN job_title IS NULL THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 1) AS job_title,
        ROUND(SUM(CASE WHEN noc16_code IS NULL THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 1) AS noc16_code,
        ROUND(SUM(CASE WHEN noc16_code_name IS NULL THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 1) AS noc16_code_name,
        ROUND(SUM(CASE WHEN noc21_code IS NULL THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 1) AS noc21_code,
        ROUND(SUM(CASE WHEN noc21_code_name IS NULL THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 1) AS noc21_code_name,
        ROUND(SUM(CASE WHEN posting_date IS NULL THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 1) AS posting_date,
        ROUND(SUM(CASE WHEN vacancy_count IS NULL THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 1) AS vacancy_count,
        ROUND(SUM(CASE WHEN education_los IS NULL THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 1) AS education_los,
        ROUND(SUM(CASE WHEN experience_level IS NULL THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 1) AS experience_level,
        ROUND(SUM(CASE WHEN naics IS NULL THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 1) AS naics,
        ROUND(SUM(CASE WHEN official_language IS NULL THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 1) AS official_language,
        ROUND(SUM(CASE WHEN province_territory IS NULL THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 1) AS province_territory,
        ROUND(SUM(CASE WHEN city IS NULL THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 1) AS city,
        ROUND(SUM(CASE WHEN economic_region IS NULL THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 1) AS economic_region,
        ROUND(SUM(CASE WHEN work_location_postal_code IS NULL THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 1) AS work_location_postal_code,
        ROUND(SUM(CASE WHEN employment_type IS NULL THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 1) AS employment_type,
        ROUND(SUM(CASE WHEN employment_term IS NULL THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 1) AS employment_term,
        ROUND(SUM(CASE WHEN employment_term_weekend IS NULL THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 1) AS employment_term_weekend,
        ROUND(SUM(CASE WHEN employment_term_telework IS NULL THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 1) AS employment_term_telework,
        ROUND(SUM(CASE WHEN salary_condition_detail IS NULL THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 1) AS salary_condition_detail,
        ROUND(SUM(CASE WHEN salary_period IS NULL THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 1) AS salary_period,
        ROUND(SUM(CASE WHEN salary_min IS NULL THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 1) AS salary_min,
        ROUND(SUM(CASE WHEN salary_max IS NULL THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 1) AS salary_max,
        ROUND(SUM(CASE WHEN hours_per IS NULL THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 1) AS hours_per,
        ROUND(SUM(CASE WHEN hours_min IS NULL THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 1) AS hours_min,
        ROUND(SUM(CASE WHEN hours_max IS NULL THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 1) AS hours_max
    FROM workspace.dev_jobmkt.stg_job_postings
)
UNPIVOT (
    null_pct FOR column_name IN (
        job_posting_id, job_title, noc16_code, noc16_code_name, noc21_code, noc21_code_name, posting_date, vacancy_count, education_los, experience_level, naics, official_language, province_territory, city, economic_region, work_location_postal_code, employment_type, employment_term, employment_term_weekend, employment_term_telework, salary_condition_detail, salary_period, salary_min, salary_max, hours_per, hours_min, hours_max
    )
)
ORDER BY null_pct DESC;
```

### Columns with High Percentage of Nulls

The columns with 50% or more than 50% null values are: education_los, experience_level, naics, employment_term_weekend,
employment_term_telework, hours_per, hours_min, hours_max, and work_location_postal_code. The next step is to check if the 57.6% observed in the columns education_los, experience_level, naics, employment_term_weekend, employment_term_telework, and hours_per is co-occurrent or independent.

```python
%sql
SELECT COUNT(*) AS rows_missing_all_together
FROM workspace.dev_jobmkt.stg_job_postings
WHERE education_los IS NULL
  AND experience_level IS NULL
  AND naics IS NULL
  AND employment_term_weekend IS NULL
  AND employment_term_telework IS NULL
  AND hours_per IS NULL;
```

## Column completeness: two posting "shapes"
education_los, experience_level, naics, employment_term_weekend,
employment_term_telework, and hours_per are all null together in
exactly 30,999 rows (57.6%). I confirmed this via a co-occurrence query.
This behavior can't be consider independent sparsity; it's structural, suggesting two
possible posting types (detailed vs. minimal). 

Decision: dropped education_los, experience_level, naics,
employment_term_weekend, employment_term_telework, and
work_location_postal_code from the model (too sparse to group by
usefully as dimension attributes). Kept hours_per, hours_min,
hours_max as fact measures despite the same null rate, since a
missing measure is a valid state, not a data-quality problem, 
and hours_per specifically enables more accurate salary
annualization for the ~42% of postings that have it.

Possible future addition: a posting_detail_level flag on
fct_postings (e.g. 'detailed' vs 'minimal') derived from whether
this null cluster is populated, so the two-path pattern is queryable
rather than just documented.
