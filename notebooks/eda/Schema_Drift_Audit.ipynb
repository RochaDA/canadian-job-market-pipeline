```python
import json
from pathlib import Path
import pandas as pd
import re
```

## Check for Missing Columns

```python
manifest_path = Path('../../data/raw/manifest.jsonl')

with open(manifest_path) as f:
    entries = [json.loads(line) for line in f if json.loads(line).get('type', 'download') == 'download']

all_columns = set()
for e in entries:
    all_columns.update(e['columns'])

for e in sorted(entries, key=lambda x: x['month']):
    missing = all_columns - set(e['columns'])
    if missing:
        print(f"{e['month']}: MISSING {sorted(missing)}")
```

## Check for difference in Date Formatting

```python

raw_dir = Path("../../data/raw")  # adjust if running from a notebook subfolder
date_pattern_slash = re.compile(r'^\d{4}/\d{2}/\d{2}$')
date_pattern_dash = re.compile(r'^\d{4}-\d{2}-\d{2}$')

for file in sorted(raw_dir.glob("job_postings_*.csv")):
    df = pd.read_csv(file, usecols=["First Posting Date"], dtype=str)
    sample = df["First Posting Date"].dropna().unique()
    slash_count = sum(1 for v in sample if date_pattern_slash.match(v))
    dash_count = sum(1 for v in sample if date_pattern_dash.match(v))
    other_count = len(sample) - slash_count - dash_count
    if other_count > 0 or (slash_count > 0 and dash_count > 0):
        print(f"{file.name}: slash={slash_count} dash={dash_count} other={other_count}")
```

## Check Final Status

```python
import json
total = 0
with open('../../data/raw/manifest.jsonl') as f:
    for line in f:
        e = json.loads(line)
        if e.get('type', 'download') == 'download':
            total += e['row_count']
print(total)
```
